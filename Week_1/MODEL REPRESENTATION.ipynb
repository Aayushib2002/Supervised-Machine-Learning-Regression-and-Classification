```python
import numpy as np #NumPy
import matplotlib.pyplot as plt
plt.style.use('./deeplearning.mplstyle')
```

```python
#x_train is the input variable
#y_train is the target
x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])
print(f"x_train = {x_train}")
print(f"y_train = {y_train}")
```

```python
#2 ways to calculate the number of training examples
#way 1
print(f"x_train.shape : {x_train.shape}")
m = x_train.shape[0]
print(f"Number of training examples: {m}")

#way 2
m = len(x_train)
print(f"Number of training examples: {m}")
```

```python
#training example x_i, y_i
i = 0
x_i = x_train[i]
y_i = y_train[i]
print(f"(x^({i}), y^({i})) = ({x_i}, {y_i})")
```

```python
#plotting data
plt.scatter(x_train, y_train, marker = 'o', c = 'g')
plt.title("Housing Prices")
plt.xlabel("Size (in 1000 sqft)")
plt.ylabel("Price (in 1000s of dollars)")
plt.show()
```

```python
#Model function
w = 200
b = 100
```

```python
#computing f(x) for data points
def compute_model_output(x, w, b):
    m = x.shape[0]
    f_wb = np.zeros(m)
    for i in range (m):
        f_wb[i] = w * x[i] + b
    return f_wb
```

```python
#calling compute_model_output function
tmp_f_wb = compute_model_output(x_train, w, b)
plt.plot(x_train, tmp_f_wb, c= 'b', label = 'Prediction')
plt.scatter(x_train, y_train, marker = 'x', c = 'y', label = 'Actual Values')
plt.title("Housing Prices")
plt.xlabel("Size (in 1000 sqft)")
plt.ylabel("Price (in 1000s of dollars)")
plt.legend()
plt.show()
```
